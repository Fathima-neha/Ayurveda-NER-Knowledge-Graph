```python
import spacy

from spacy.tokens import DocBin
```

```python
import pandas as pd

df = pd.read_excel("/content/annotations.xlsx")
```

```python
TRAIN_DATA = []

for sentence in df["Sentence"].unique():

    temp = df[df["Sentence"]==sentence]

    entities=[]

    for _,row in temp.iterrows():

        start=sentence.find(row["Entity"])

        end=start+len(row["Entity"])

        entities.append((start,end,row["Label"]))

    TRAIN_DATA.append((sentence,{"entities":entities}))
```

```python
nlp = spacy.blank("en")

db = DocBin()

for text, annot in TRAIN_DATA:

    doc = nlp.make_doc(text)

    ents=[]

    for start,end,label in annot["entities"]:

        span=doc.char_span(start,end,label=label)

        if span is not None:

            ents.append(span)

    doc.ents=ents

    db.add(doc)

db.to_disk("train.spacy")
```
